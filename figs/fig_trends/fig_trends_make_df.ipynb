```python
from pathlib import Path
import pandas as pd

df_path = Path('../../dfs/df_trends.csv')
df_path.parent
```

```python

defaults = {'duration':11, 'snr': 10, 'sample_freq':64}
variants = dict(
    duration = [3, 7, 15],
    sample_freq = [8, 16, 32, 128, 256],
    snr = [0.1, 1, 100, 1000],
)
num_samples_each = 100

configs = []
for key, vals in variants.items():
    for val in vals:
        config = {**defaults, key: val}
        config['tag'] = f'{key}_{val}'
        configs.append(config)

rows = []
for i in range(num_samples_each):
    for config in configs:
        rows.append({**config, 'pid':f'p{i:03d}'})
df = pd.DataFrame(rows)
df
```

```python
df.to_csv(df_path, index=False)
```
